```python
diabetes = datasets.load_diabetes()
x_reg = diabetes.data[:, np.newaxis,2]
y_reg = diabetes.target

iris = datasets.load_iris()
X_clf = iris.data[:, :2]
Y_clf = iris.target

print("Dataset Berhasil Anjing")
```

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))


ax[0].scatter(x_reg, y_reg, color='blue', alpha=0.5)
ax[0].set_title('Scatterplot')
ax[0].set_xlabel('Bmi')
ax[0].set_ylabel('Tingkat Dewa Slot')

scatter = ax[1].scatter(X_clf[:, 0], X_clf[:, 1], c=Y_clf, cmap='brg', edgecolor='k')
ax[1].set_title('Pola Pikir Lu Dimana Wok')
ax[1].set_xlabel('Sepal Length')
ax[1].set_ylabel('Sepal Width')

plt.tight_layout()
plt.show()
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(x_reg,y_reg, test_size=0.2, random_state=42)

model_reg = LinearRegression()
model_reg.fit(X_train_reg, y_train_reg)

y_pred_reg = model_reg.predict(X_test_reg)

plt.figure(figsize=(6, 4))
plt.scatter(X_test_reg, y_test_reg, color='blue', alpha=0.6, label='Data Asli')
plt.plot(X_test_reg, y_pred_reg, color='red', linewidth=2, label='Regresi Linier')
plt.title('Regresi Linier Anjing')
plt.legend()
plt.show()

berat_badan = 65
tinggi_badan = 167
bmi = berat_badan / ((tinggi_badan) ** 2)
bmi_baru = np.array([[bmi]])
prediksi_diabetes = model_reg.predict(bmi_baru)

print(f"Prediksi Parlay {bmi * 100} adalah {prediksi_diabetes[0]}")
```
